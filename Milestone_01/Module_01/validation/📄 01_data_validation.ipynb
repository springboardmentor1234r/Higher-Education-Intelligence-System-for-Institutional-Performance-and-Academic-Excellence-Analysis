```python
import pandas as pd
from pathlib import Path

# Project folders
RAW_DIR = Path("../data/raw")
VALIDATION_DIR = Path("../validation")

VALIDATION_DIR.mkdir(parents=True, exist_ok=True)

print("Raw data folder:", RAW_DIR.resolve())
print("Validation folder:", VALIDATION_DIR.resolve())
```

```python
# ============================================================
# LOAD ALL FOUR DATA SOURCES
# ============================================================

# QS 2025
qs_df = pd.read_csv(
    RAW_DIR / "qs_2025_raw.csv"
)

# THE 2024
the_df = pd.read_csv(
    RAW_DIR / "the_2024_raw.csv"
)

# WUR 2023
wur_df = pd.read_csv(
    RAW_DIR / "World University Rankings 2023.csv"
)

# World Bank Education Statistics
wb_country_series_df = pd.read_csv(
    RAW_DIR / "EdStatsCountry-Series.csv"
)

wb_country_df = pd.read_csv(
    RAW_DIR / "EdStatsCountry.csv"
)

wb_footnote_df = pd.read_csv(
    RAW_DIR / "EdStatsFootNote.csv"
)

wb_series_df = pd.read_csv(
    RAW_DIR / "EdStatsSeries.csv"
)

# World Bank Excel workbook
wb_excel = pd.ExcelFile(
    RAW_DIR / "EdStatsEXCEL.xlsx"
)

print("All datasets loaded successfully!")
```

```python
print("QS 2025:", qs_df.shape)
print("THE 2024:", the_df.shape)
print("WUR 2023:", wur_df.shape)

print("World Bank Country-Series:", wb_country_series_df.shape)
print("World Bank Country:", wb_country_df.shape)
print("World Bank FootNote:", wb_footnote_df.shape)
print("World Bank Series:", wb_series_df.shape)

print("World Bank Excel sheets:", wb_excel.sheet_names)
```

```python
from pathlib import Path
import pandas as pd

RAW_DIR = Path("../data/raw")

print("RAW DATA FILES")
print("=" * 60)

for file in RAW_DIR.rglob("*"):
    if file.is_file():
        print(file.relative_to(RAW_DIR))
```

```python
print("\nDATASET COLUMNS")
print("=" * 60)

for file in RAW_DIR.rglob("*"):

    if not file.is_file():
        continue

    try:

        if file.suffix.lower() == ".csv":
            df = pd.read_csv(file, nrows=5)

        elif file.suffix.lower() in [".xlsx", ".xls"]:
            # Only inspect the workbook structure; don't load the whole thing
            excel = pd.ExcelFile(file)
            print(f"\n{file.name}")
            print("Excel sheets:", excel.sheet_names)
            continue

        else:
            continue

        print(f"\n{file.name}")
        print("Rows sampled:", len(df))
        print("Columns:")
        print(list(df.columns))

    except Exception as e:
        print(f"\nCould not inspect {file.name}: {e}")
```
