```python
import pandas as pd
import numpy as np
```

```python
qs = pd.read_csv("QS World University Rankings 2025 (Top global universities).csv", encoding='latin1')

the = pd.read_csv("TIMES_WorldUniversityRankings_2024.csv", encoding='latin1')

wur = pd.read_csv("World University Rankings 2023.csv", encoding='latin1')
```

```python
datasets = {
    "QS": qs,
    "THE": the,
    "WUR": wur
}

for name, df in datasets.items():

    print("="*60)
    print(name)
    print("="*60)

    print("Rows :", df.shape[0])
    print("Columns :", df.shape[1])

    print("\nColumn Names:\n")

    print(df.columns.tolist())

    print("\n")
```

```python
for name, df in datasets.items():

    print("\n")
    print("="*60)
    print(name)
    print("="*60)

    missing = pd.DataFrame({
        "Missing Count": df.isnull().sum(),
        "Missing %": round(
            (df.isnull().sum()/len(df))*100,
            2
        )
    })

    print(
        missing.sort_values(
            "Missing %",
            ascending=False
        ).head(20)
    )
```

```python
for name, df in datasets.items():

    print("\n")
    print("="*60)
    print(name)
    print("="*60)

    print(
        "Duplicate Rows:",
        df.duplicated().sum()
    )
```

```python
def clean_columns(df):

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
        .str.replace("/", "_")
    )

    return df

qs = clean_columns(qs)

the = clean_columns(the)

wur = clean_columns(wur)
```

```python
print(qs.columns.tolist())

print(the.columns.tolist())

print(wur.columns.tolist())
```

```python
qs.to_csv(
    "qs_columns_cleaned.csv",
    index=False
)

the.to_csv(
    "the_columns_cleaned.csv",
    index=False
)

wur.to_csv(
    "wur_columns_cleaned.csv",
    index=False
)
```

```python
print(qs.columns.tolist())
print(the.columns.tolist())
print(wur.columns.tolist())
```

```python
import re

def clean_name(x):

    x = str(x).lower()

    x = re.sub(r'[^\w\s]', '', x)

    x = re.sub(r'\s+', ' ', x)

    return x.strip()
```

```python
qs["university_name"] = qs["institution_name"].apply(clean_name)

the["university_name"] = the["name"].apply(clean_name)

wur["university_name"] = wur["name_of_university"].apply(clean_name)
```

```python
def clean_country(x):

    x = str(x).lower().strip()

    country_map = {

        "usa":"united states",
        "us":"united states",
        "u.s.a":"united states",

        "uk":"united kingdom",
        "u.k":"united kingdom"
    }

    return country_map.get(x,x)
```

```python
qs["country"] = qs["location"].apply(clean_country)

wur["country"] = wur["location"].apply(clean_country)
```

```python
qs = qs.drop_duplicates()

the = the.drop_duplicates()

wur = wur.drop_duplicates()
```

```python
for name, df in {

    "QS": qs,
    "THE": the,
    "WUR": wur

}.items():

    print("\n",name)

    print(df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
qs.to_csv(
    "qs_cleaned.csv",
    index=False
)

the.to_csv(
    "the_cleaned.csv",
    index=False
)

wur.to_csv(
    "wur_cleaned.csv",
    index=False
)
```
