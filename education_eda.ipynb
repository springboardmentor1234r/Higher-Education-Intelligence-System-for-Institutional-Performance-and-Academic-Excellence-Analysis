```python
# ============================================================
# EDUCATION EDA - COMPLETE ANALYSIS
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv("../data/university_cleaned_data.csv")

print("Dataset Shape:", df.shape)


# ============================================================
# 1. TOP 10 UNIVERSITIES BY QS RANK
# ============================================================

print("\n========== TOP 10 UNIVERSITIES BY QS RANK ==========")

top_qs = df.sort_values("QS_Rank").head(10)

display(
    top_qs[
        [
            "University_Name_QS",
            "Country_QS",
            "QS_Rank",
            "QS_Overall_Score"
        ]
    ]
)


# ============================================================
# 2. TOP 10 UNIVERSITIES BY THE RANK
# ============================================================

print("\n========== TOP 10 UNIVERSITIES BY THE RANK ==========")

top_the = df.sort_values("THE_Rank").head(10)

display(
    top_the[
        [
            "University_Name_THE",
            "Country_THE",
            "THE_Rank",
            "THE_Overall_Score"
        ]
    ]
)


# ============================================================
# 3. QS VS THE RANKING COMPARISON
# ============================================================

print("\n========== QS VS THE RANKING ==========")

comparison = df[
    [
        "University_Name_QS",
        "QS_Rank",
        "THE_Rank",
        "QS_Overall_Score"
    ]
].copy()

display(comparison.head(10))


# ============================================================
# 4. TOP COUNTRIES BY NUMBER OF UNIVERSITIES
# ============================================================

print("\n========== TOP COUNTRIES ==========")

country_counts = df["Country_QS"].value_counts().head(10)

display(country_counts)


# Chart

plt.figure(figsize=(10, 5))

country_counts.plot(kind="bar")

plt.title("Top 10 Countries by Number of Universities")
plt.xlabel("Country")
plt.ylabel("Number of Universities")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 5. AVERAGE QS SCORE BY COUNTRY
# ============================================================

print("\n========== AVERAGE QS SCORE BY COUNTRY ==========")

country_avg_score = (
    df.groupby("Country_QS")["QS_Overall_Score"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

display(country_avg_score)


# Chart

plt.figure(figsize=(10, 5))

country_avg_score.plot(kind="bar")

plt.title("Top 10 Countries by Average QS Score")
plt.xlabel("Country")
plt.ylabel("Average QS Score")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 6. QS SCORE DISTRIBUTION
# ============================================================

print("\n========== QS SCORE DISTRIBUTION ==========")

plt.figure(figsize=(8, 5))

plt.hist(df["QS_Overall_Score"], bins=20)

plt.title("Distribution of QS Overall Scores")
plt.xlabel("QS Overall Score")
plt.ylabel("Number of Universities")
plt.tight_layout()
plt.show()


# ============================================================
# 7. TOP UNIVERSITIES BY QS OVERALL SCORE
# ============================================================

print("\n========== TOP 10 BY QS OVERALL SCORE ==========")

top_score = df.sort_values(
    "QS_Overall_Score",
    ascending=False
).head(10)

display(
    top_score[
        [
            "University_Name_QS",
            "Country_QS",
            "QS_Rank",
            "QS_Overall_Score"
        ]
    ]
)


# ============================================================
# 8. STUDENT POPULATION
# ============================================================

print("\n========== STUDENT POPULATION ==========")

largest_universities = df.sort_values(
    "Total_students",
    ascending=False
).head(10)

display(
    largest_universities[
        [
            "University_Name_QS",
            "Country_QS",
            "Total_students",
            "Student_Staff_Ratio"
        ]
    ]
)


# ============================================================
# 9. STUDENT-STAFF RATIO
# ============================================================

print("\n========== STUDENT-STAFF RATIO ==========")

print("Average Student-Staff Ratio:",
      round(df["Student_Staff_Ratio"].mean(), 2))

print("Minimum:",
      round(df["Student_Staff_Ratio"].min(), 2))

print("Maximum:",
      round(df["Student_Staff_Ratio"].max(), 2))


# ============================================================
# 10. QS ACADEMIC REPUTATION
# ============================================================

print("\n========== ACADEMIC REPUTATION ==========")

academic = df.sort_values(
    "QS_Academic_Reputation",
    ascending=False
).head(10)

display(
    academic[
        [
            "University_Name_QS",
            "Country_QS",
            "QS_Academic_Reputation"
        ]
    ]
)


# ============================================================
# 11. QS EMPLOYER REPUTATION
# ============================================================

print("\n========== EMPLOYER REPUTATION ==========")

employer = df.sort_values(
    "QS_Employer_Reputation",
    ascending=False
).head(10)

display(
    employer[
        [
            "University_Name_QS",
            "Country_QS",
            "QS_Employer_Reputation"
        ]
    ]
)


# ============================================================
# 12. QS FACULTY-STUDENT SCORE
# ============================================================

print("\n========== FACULTY-STUDENT SCORE ==========")

faculty = df.sort_values(
    "QS_Faculty_Student_Score",
    ascending=False
).head(10)

display(
    faculty[
        [
            "University_Name_QS",
            "Country_QS",
            "QS_Faculty_Student_Score"
        ]
    ]
)


# ============================================================
# 13. INTERNATIONAL FACULTY
# ============================================================

print("\n========== INTERNATIONAL FACULTY ==========")

international_faculty = df.sort_values(
    "QS_International_Faculty",
    ascending=False
).head(10)

display(
    international_faculty[
        [
            "University_Name_QS",
            "Country_QS",
            "QS_International_Faculty"
        ]
    ]
)


# ============================================================
# 14. INTERNATIONAL STUDENTS
# ============================================================

print("\n========== INTERNATIONAL STUDENTS ==========")

international_students = df.sort_values(
    "QS_International_Students",
    ascending=False
).head(10)

display(
    international_students[
        [
            "University_Name_QS",
            "Country_QS",
            "QS_International_Students"
        ]
    ]
)


# ============================================================
# 15. QS RANK VS QS SCORE
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["QS_Rank"],
    df["QS_Overall_Score"],
    alpha=0.6
)

plt.title("QS Rank vs QS Overall Score")
plt.xlabel("QS Rank")
plt.ylabel("QS Overall Score")
plt.tight_layout()
plt.show()


# ============================================================
# 16. ACADEMIC REPUTATION VS OVERALL SCORE
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["QS_Academic_Reputation"],
    df["QS_Overall_Score"],
    alpha=0.6
)

plt.title("Academic Reputation vs QS Overall Score")
plt.xlabel("Academic Reputation")
plt.ylabel("QS Overall Score")
plt.tight_layout()
plt.show()


# ============================================================
# 17. STUDENT-STAFF RATIO VS QS SCORE
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Student_Staff_Ratio"],
    df["QS_Overall_Score"],
    alpha=0.6
)

plt.title("Student-Staff Ratio vs QS Overall Score")
plt.xlabel("Student-Staff Ratio")
plt.ylabel("QS Overall Score")
plt.tight_layout()
plt.show()


# ============================================================
# 18. CORRELATION ANALYSIS
# ============================================================

print("\n========== CORRELATION WITH QS OVERALL SCORE ==========")

numeric_columns = df.select_dtypes(
    include="number"
).columns

correlation = (
    df[numeric_columns]
    .corr()["QS_Overall_Score"]
    .sort_values(ascending=False)
)

display(correlation)


# ============================================================
# 19. YEAR-WISE UNIVERSITY COUNT
# ============================================================

print("\n========== YEAR-WISE UNIVERSITY COUNT ==========")

year_counts = df["Year"].value_counts().sort_index()

display(year_counts)

plt.figure(figsize=(8, 5))

year_counts.plot(kind="bar")

plt.title("Universities by Year")
plt.xlabel("Year")
plt.ylabel("Number of Universities")
plt.tight_layout()
plt.show()


# ============================================================
# 20. FINAL EDA SUMMARY
# ============================================================

print("\n========================================")
print("           EDA COMPLETED")
print("========================================")

print("Total Universities:", len(df))
print("Total Countries:", df["Country_QS"].nunique())
print(
    "Average QS Score:",
    round(df["QS_Overall_Score"].mean(), 2)
)
print(
    "Best QS Rank:",
    int(df["QS_Rank"].min())
)
print(
    "Average Student-Staff Ratio:",
    round(df["Student_Staff_Ratio"].mean(), 2)
)

print("\nTop Country by University Count:")
print(df["Country_QS"].value_counts().idxmax())

print("\nTop University by QS Rank:")
print(
    df.sort_values("QS_Rank")
    .iloc[0]["University_Name_QS"]
)

print("\n========================================")
```

```python
# ==========================================
# COMPLETE EDA - EduVision Dataset
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# Load cleaned dataset again
df = pd.read_csv("../data/university_cleaned_data.csv")

print("Dataset shape:", df.shape)

# ------------------------------------------
# 1. Check columns
# ------------------------------------------

print("\nColumns:")
print(df.columns.tolist())


# ------------------------------------------
# 2. Top universities by QS ranking
# ------------------------------------------

print("\n========== TOP 10 QS UNIVERSITIES ==========")

top_qs = df.sort_values("QS_Rank", ascending=True).head(10)

print(
    top_qs[
        ["University_Name_QS", "Country_QS", "QS_Rank", "QS_Overall_Score"]
    ].to_string(index=False)
)


# ------------------------------------------
# 3. Top universities by THE ranking
# ------------------------------------------

print("\n========== TOP 10 THE UNIVERSITIES ==========")

top_the = df.sort_values("THE_Rank", ascending=True).head(10)

print(
    top_the[
        ["University_Name_THE", "Country_THE", "THE_Rank", "THE_Overall_Score"]
    ].to_string(index=False)
)


# ------------------------------------------
# 4. Universities by country
# ------------------------------------------

print("\n========== UNIVERSITIES BY COUNTRY ==========")

country_counts = df["Country_QS"].value_counts().head(15)

print(country_counts)


# ------------------------------------------
# 5. Average QS score by country
# ------------------------------------------

print("\n========== AVERAGE QS SCORE BY COUNTRY ==========")

country_qs_score = (
    df.groupby("Country_QS")["QS_Overall_Score"]
    .mean()
    .sort_values(ascending=False)
    .head(15)
)

print(country_qs_score)


# ------------------------------------------
# 6. Student population
# ------------------------------------------

print("\n========== STUDENT POPULATION ==========")

# Check whether the column exists
if "Total_students" in df.columns:

    largest_universities = (
        df.sort_values("Total_students", ascending=False)
        .head(10)
    )

    print(
        largest_universities[
            ["University_Name_QS", "Country_QS", "Total_students"]
        ].to_string(index=False)
    )

else:
    print("Total_students column not available.")


# ------------------------------------------
# 7. QS ranking distribution
# ------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(df["QS_Rank"], bins=20)

plt.title("Distribution of QS Rankings")
plt.xlabel("QS Rank")
plt.ylabel("Number of Universities")

plt.show()


# ------------------------------------------
# 8. QS overall score distribution
# ------------------------------------------

plt.figure(figsize=(10, 5))

plt.hist(df["QS_Overall_Score"], bins=20)

plt.title("Distribution of QS Overall Scores")
plt.xlabel("QS Overall Score")
plt.ylabel("Number of Universities")

plt.show()


# ------------------------------------------
# 9. Top 10 countries by university count
# ------------------------------------------

plt.figure(figsize=(10, 5))

country_counts.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Countries by Number of Universities")
plt.xlabel("Number of Universities")
plt.ylabel("Country")

plt.show()


# ------------------------------------------
# 10. QS vs THE overall score
# ------------------------------------------

plt.figure(figsize=(8, 6))

plt.scatter(
    df["QS_Overall_Score"],
    df["THE_Overall_Score"],
    alpha=0.6
)

plt.title("QS Overall Score vs THE Overall Score")
plt.xlabel("QS Overall Score")
plt.ylabel("THE Overall Score")

plt.show()


# ------------------------------------------
# 11. Correlation between ranking metrics
# ------------------------------------------

print("\n========== CORRELATION ==========")

numeric_columns = [
    "QS_Rank",
    "QS_Overall_Score",
    "QS_Academic_Reputation",
    "QS_Employer_Reputation",
    "QS_Faculty_Student_Score",
    "QS_Citations_Per_Faculty",
    "QS_International_Faculty",
    "QS_International_Students",
    "THE_Rank",
    "THE_Overall_Score",
    "THE_Teaching_Score",
    "THE_Research_Score",
    "THE_Citations_Score",
    "THE_Industry_Income_Score",
    "THE_International_Outlook"
]

# Keep only columns that actually exist
available_columns = [
    col for col in numeric_columns
    if col in df.columns
]

correlation = df[available_columns].corr()

print(correlation.round(2))


# ------------------------------------------
# 12. Basic summary
# ------------------------------------------

print("\n========== FINAL SUMMARY ==========")

print("Number of universities:", len(df))
print("Number of columns:", len(df.columns))

print("\nCountries represented:")
print(df["Country_QS"].nunique())

print("\nYears:")
print(df["Year"].unique())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nEDA completed successfully!")
```
