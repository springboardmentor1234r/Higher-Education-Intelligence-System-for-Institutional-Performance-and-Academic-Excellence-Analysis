**Milestone 1: Module 2: Data Cleaning & Transformation**

This notebook cleans university and country entities, eliminates duplicate records, handles missing indicators responsibly, and converts numeric metrics.

```python
import os
import re
import pandas as pd
import numpy as np

# Load raw files
qs = pd.read_csv('01_raw_datasets/qs_2025_raw.csv', encoding='latin1')
the = pd.read_csv('01_raw_datasets/the_2024_raw.csv', encoding='latin1')
wur = pd.read_csv('01_raw_datasets/wur_2023_raw.csv', encoding='latin1')

# Deduplicate primary ranking
qs = qs.drop_duplicates(subset=['Institution_Name', 'Location'])
print(f'Deduplicated QS Institutions: {len(qs)}')
```

## Entity Normalization & Standard Name Cleaning
Per Section 11 & 12 of the project guide, we remove parenthetical acronyms and punctuation, lowercase strings, and normalize whitespace.

```python
def clean_uni(name):
    if pd.isna(name): return ''
    s = re.sub(r'\(.*?\)', '', str(name))
    s = re.sub(r'[^\w\s]', '', s)
    return re.sub(r'\s+', ' ', s).strip().lower()

qs['clean_uni'] = qs['Institution_Name'].apply(clean_uni)
the['clean_uni'] = the['name'].apply(clean_uni)
wur['clean_uni'] = wur['Name of University'].apply(clean_uni)
print('Standardization complete.')
```
