```python
import re

def extract_rank_start(value):
    if pd.isna(value):
        return pd.NA

    # Remove "=" and spaces
    value = str(value).replace("=", "").strip()

    # Extract the first number
    match = re.search(r"\d+", value)

    if match:
        return int(match.group())

    return pd.NA


df["QS_Rank_Numeric"] = df["QS_Rank"].apply(extract_rank_start)
df["THE_Rank_Numeric"] = df["THE_Rank"].apply(extract_rank_start)

df["QS_Rank_Numeric"] = pd.to_numeric(
    df["QS_Rank_Numeric"],
    errors="coerce"
)

df["THE_Rank_Numeric"] = pd.to_numeric(
    df["THE_Rank_Numeric"],
    errors="coerce"
)
```

```python
def convert_the_score(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip()

    # Handle ranges such as 55.9–58.6
    if "–" in value:
        parts = value.split("–")
        try:
            lower = float(parts[0])
            upper = float(parts[1])
            return (lower + upper) / 2
        except:
            return pd.NA

    # Handle normal numeric values
    try:
        return float(value)
    except:
        return pd.NA


df["THE_Overall_Score_Numeric"] = (
    df["THE_Overall_Score"].apply(convert_the_score)
)

df["THE_Overall_Score_Numeric"] = pd.to_numeric(
    df["THE_Overall_Score_Numeric"],
    errors="coerce"
)
```
