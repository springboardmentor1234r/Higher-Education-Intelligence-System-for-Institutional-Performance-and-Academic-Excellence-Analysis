```python
import pandas as pd
import os
import re

# Load the cleaned datasets
qs_clean = pd.read_csv(
    os.path.join(CLEANED_PATH, "QS_2025_Cleaned.csv")
)

the_clean = pd.read_csv(
    os.path.join(CLEANED_PATH, "THE_2024_Cleaned.csv")
)

wur_clean = pd.read_csv(
    os.path.join(CLEANED_PATH, "WUR_2023_Cleaned.csv")
)

# Function to standardize university names
def standardize_university_name(series):
    return (
        series.astype("string")
        .str.strip()
        .str.lower()
        .str.replace(r"[^\w\s]", "", regex=True)
        .str.replace(r"\s+", " ", regex=True)
    )

# Create standardized name columns
qs_clean["university_name_std"] = standardize_university_name(
    qs_clean["Institution_Name"]
)

the_clean["university_name_std"] = standardize_university_name(
    the_clean["name"]
)

wur_clean["university_name_std"] = standardize_university_name(
    wur_clean["Name of University"]
)

print("===== UNIVERSITY NAME STANDARDIZATION =====")
print("QS 2025:", qs_clean["university_name_std"].notna().sum(), "names standardized")
print("THE 2024:", the_clean["university_name_std"].notna().sum(), "names standardized")
print("WUR 2023:", wur_clean["university_name_std"].notna().sum(), "names standardized")

print("\nExamples:")
print("\nQS:")
print(qs_clean[["Institution_Name", "university_name_std"]].dropna().head(5).to_string(index=False))

print("\nTHE:")
print(the_clean[["name", "university_name_std"]].dropna().head(5).to_string(index=False))

print("\nWUR:")
print(wur_clean[["Name of University", "university_name_std"]].dropna().head(5).to_string(index=False))
```

```python
# World Bank country reference
country_reference = (
    wb_country_info[["Country Code", "Short Name"]]
    .dropna()
    .drop_duplicates()
    .copy()
)

# Create a lookup using lowercase country names
country_lookup = {
    name.strip().lower(): name
    for name in country_reference["Short Name"]
}

# Common naming differences found in ranking datasets
country_aliases = {
    "usa": "United States",
    "united states of america": "United States",
    "uk": "United Kingdom",
    "u.k.": "United Kingdom",
    "russia": "Russian Federation",
    "south korea": "Korea, Rep.",
    "republic of korea": "Korea, Rep.",
    "iran": "Iran, Islamic Rep.",
    "vietnam": "Viet Nam",
    "venezuela": "Venezuela, RB",
    "bolivia": "Bolivia",
    "tanzania": "Tanzania",
    "moldova": "Moldova",
    "laos": "Lao PDR",
    "egypt": "Egypt, Arab Rep."
}

def standardize_country_name(series):
    result = series.astype("string").str.strip()

    result = result.map(
        lambda x: (
            pd.NA
            if pd.isna(x)
            else country_aliases.get(
                x.lower(),
                country_lookup.get(x.lower(), x)
            )
        )
    )

    return result

# Standardize country names
qs_clean["country_name_std"] = standardize_country_name(
    qs_clean["Location"]
)

the_clean["country_name_std"] = standardize_country_name(
    the_clean["location"]
)

wur_clean["country_name_std"] = standardize_country_name(
    wur_clean["Location"]
)

print("===== COUNTRY NAME STANDARDIZATION =====")
print("QS countries:", qs_clean["country_name_std"].nunique())
print("THE countries:", the_clean["country_name_std"].nunique())
print("WUR countries:", wur_clean["country_name_std"].nunique())

print("\nExamples:")
print("\nQS:")
print(
    qs_clean[["Location", "country_name_std"]]
    .drop_duplicates()
    .head(10)
    .to_string(index=False)
)

print("\nTHE:")
print(
    the_clean[["location", "country_name_std"]]
    .drop_duplicates()
    .head(10)
    .to_string(index=False)
)

print("\nWUR:")
print(
    wur_clean[["Location", "country_name_std"]]
    .drop_duplicates()
    .head(10)
    .to_string(index=False)
)
```

```python
# Create a master list of unique university + country combinations
university_master = pd.concat([
    qs_clean[["university_name_std", "country_name_std"]],
    the_clean[["university_name_std", "country_name_std"]],
    wur_clean[["university_name_std", "country_name_std"]]
], ignore_index=True)

# Remove records where university name is missing
university_master = university_master.dropna(
    subset=["university_name_std"]
).drop_duplicates()

# Create a stable University ID
university_master = university_master.sort_values(
    ["university_name_std", "country_name_std"],
    na_position="last"
).reset_index(drop=True)

university_master["university_id"] = [
    f"U{i:05d}" for i in range(1, len(university_master) + 1)
]

# Create lookup key
university_master["university_key"] = (
    university_master["university_name_std"]
    + "||"
    + university_master["country_name_std"].fillna("")
)

# Apply IDs to each dataset
def add_university_id(df):
    df["university_key"] = (
        df["university_name_std"]
        + "||"
        + df["country_name_std"].fillna("")
    )

    return df.merge(
        university_master[["university_key", "university_id"]],
        on="university_key",
        how="left"
    )

qs_clean = add_university_id(qs_clean)
the_clean = add_university_id(the_clean)
wur_clean = add_university_id(wur_clean)

print("===== UNIVERSITY ID CREATION =====")
print("Unique universities in master:", len(university_master))

print("\nQS IDs created:",
      qs_clean["university_id"].notna().sum())

print("THE IDs created:",
      the_clean["university_id"].notna().sum())

print("WUR IDs created:",
      wur_clean["university_id"].notna().sum())

print("\nMissing University IDs:")
print("QS:", qs_clean["university_id"].isna().sum())
print("THE:", the_clean["university_id"].isna().sum())
print("WUR:", wur_clean["university_id"].isna().sum())

print("\nSample University IDs:")
print(
    university_master[
        ["university_id", "university_name_std", "country_name_std"]
    ].head(10).to_string(index=False)
)
```

```python
# Create country master from World Bank country information
country_master = (
    wb_country_info[["Country Code", "Short Name"]]
    .dropna(subset=["Country Code", "Short Name"])
    .drop_duplicates()
    .copy()
)

# Create Country_ID
country_master = country_master.sort_values("Country Code").reset_index(drop=True)

country_master["country_id"] = [
    f"C{i:03d}" for i in range(1, len(country_master) + 1)
]

# Create lookup from standardized country name to Country_ID
country_id_lookup = dict(
    zip(
        country_master["Short Name"].str.strip().str.lower(),
        country_master["country_id"]
    )
)

# Add Country_ID to university master
university_master["country_id"] = (
    university_master["country_name_std"]
    .astype("string")
    .str.strip()
    .str.lower()
    .map(country_id_lookup)
)

print("===== COUNTRY ID CREATION =====")
print("Countries in master:", len(country_master))

print("\nUniversity master records:", len(university_master))
print(
    "University records with Country_ID:",
    university_master["country_id"].notna().sum()
)
print(
    "University records missing Country_ID:",
    university_master["country_id"].isna().sum()
)

print("\nSample Country IDs:")
print(
    country_master[
        ["country_id", "Country Code", "Short Name"]
    ].head(10).to_string(index=False)
)
```

```python
# Explicit mapping to World Bank country codes
country_code_map = {
    "Russian Federation": "RUS",
    "Iran, Islamic Rep.": "IRN",
    "Korea, Rep.": "KOR",
    "Egypt, Arab Rep.": "EGY",
    "Venezuela, RB": "VEN",
    "Viet Nam": "VNM",
    "Brunei Darussalam": "BRN",
    "North Macedonia": "MKD"
}

# Create Country Code -> Country_ID lookup
code_to_id = dict(
    zip(
        country_master["Country Code"],
        country_master["country_id"]
    )
)

# Start with existing Country_ID values
university_master["country_id"] = (
    university_master["country_id"]
)

# Apply explicit World Bank code mappings
for country_name, country_code in country_code_map.items():
    mask = university_master["country_name_std"] == country_name
    university_master.loc[mask, "country_id"] = code_to_id.get(country_code)

print("===== FINAL COUNTRY ID CHECK =====")
print("Total university records:", len(university_master))
print(
    "Records with Country_ID:",
    university_master["country_id"].notna().sum()
)
print(
    "Records still missing Country_ID:",
    university_master["country_id"].isna().sum()
)

print("\nRemaining unmatched country names:")
print(
    university_master[
        university_master["country_id"].isna()
    ]["country_name_std"]
    .value_counts(dropna=False)
    .to_string()
)
```

```python
import os
import pandas as pd
import numpy as np

# ============================================================
# PATHS
# ============================================================

PROJECT_PATH = "/content/drive/MyDrive/EduVision_DV"
CLEANED_PATH = os.path.join(PROJECT_PATH, "02_Cleaned_Data")
PROCESSED_PATH = os.path.join(PROJECT_PATH, "03_Processed_Data")

# ============================================================
# LOAD SAVED DATA
# ============================================================

dim_university = pd.read_csv(
    os.path.join(PROCESSED_PATH, "dim_university.csv")
)

performance = pd.read_csv(
    os.path.join(PROCESSED_PATH, "fact_university_performance.csv")
)

student = pd.read_csv(
    os.path.join(PROCESSED_PATH, "fact_student.csv")
)

qs = pd.read_csv(
    os.path.join(CLEANED_PATH, "QS_2025_Cleaned.csv")
)

the = pd.read_csv(
    os.path.join(CLEANED_PATH, "THE_2024_Cleaned.csv")
)

# ============================================================
# BASE DATASET
# ============================================================

final_df = dim_university[
    [
        "university_id",
        "university_name",
        "country_id",
        "country_name",
        "region"
    ]
].copy()

# ============================================================
# KPI 1 + KPI 5
# QS 2025
# ============================================================

performance_kpi = performance[
    [
        "university_id",
        "overall_score",
        "academic_reputation"
    ]
].copy()

performance_kpi["overall_score"] = pd.to_numeric(
    performance_kpi["overall_score"],
    errors="coerce"
)

performance_kpi["academic_reputation"] = pd.to_numeric(
    performance_kpi["academic_reputation"],
    errors="coerce"
)

performance_kpi = performance_kpi.drop_duplicates(
    "university_id"
)

final_df = final_df.merge(
    performance_kpi,
    on="university_id",
    how="left"
)

final_df.rename(
    columns={
        "overall_score": "global_ranking_score",
        "academic_reputation": "academic_reputation_score"
    },
    inplace=True
)

# ============================================================
# KPI 2
# QS Citations per Faculty
# ============================================================

def standardize_name(series):
    return (
        series.astype("string")
        .str.strip()
        .str.lower()
        .str.replace(r"[^\w\s]", "", regex=True)
        .str.replace(r"\s+", " ", regex=True)
    )

dim_match = dim_university[
    ["university_id", "university_name"]
].copy()

dim_match["name_std"] = standardize_name(
    dim_match["university_name"]
)

qs["name_std"] = standardize_name(
    qs["Institution_Name"]
)

qs_kpi2 = qs[
    [
        "name_std",
        "Citations_per_Faculty_Score"
    ]
].copy()

qs_kpi2["Citations_per_Faculty_Score"] = pd.to_numeric(
    qs_kpi2["Citations_per_Faculty_Score"],
    errors="coerce"
)

qs_kpi2 = qs_kpi2.drop_duplicates("name_std")

qs_kpi2 = dim_match.merge(
    qs_kpi2,
    on="name_std",
    how="left"
)

qs_kpi2 = qs_kpi2[
    [
        "university_id",
        "Citations_per_Faculty_Score"
    ]
]

final_df = final_df.merge(
    qs_kpi2,
    on="university_id",
    how="left"
)

final_df.rename(
    columns={
        "Citations_per_Faculty_Score": "research_impact_score"
    },
    inplace=True
)

# ============================================================
# KPI 3 + KPI 4
# WUR 2023
# ============================================================

student_kpi = student[
    [
        "university_id",
        "students_per_staff",
        "international_student_percentage"
    ]
].copy()

student_kpi["students_per_staff"] = pd.to_numeric(
    student_kpi["students_per_staff"],
    errors="coerce"
)

student_kpi["international_student_percentage"] = pd.to_numeric(
    student_kpi["international_student_percentage"],
    errors="coerce"
)

student_kpi = student_kpi.drop_duplicates(
    "university_id"
)

final_df = final_df.merge(
    student_kpi,
    on="university_id",
    how="left"
)

final_df.rename(
    columns={
        "students_per_staff": "faculty_to_student_ratio"
    },
    inplace=True
)

# ============================================================
# KPI 6
# THE 2024
# 50% Research + 30% Citations + 20% International Outlook
# ============================================================

the["name_std"] = standardize_name(
    the["name"]
)

the_kpi6 = the[
    [
        "name_std",
        "scores_research",
        "scores_citations",
        "scores_international_outlook"
    ]
].copy()

for column in [
    "scores_research",
    "scores_citations",
    "scores_international_outlook"
]:
    the_kpi6[column] = pd.to_numeric(
        the_kpi6[column],
        errors="coerce"
    )

the_kpi6 = the_kpi6.drop_duplicates(
    "name_std"
)

the_kpi6 = dim_match.merge(
    the_kpi6,
    on="name_std",
    how="left"
)

the_kpi6["research_productivity_index"] = (
    0.50 * the_kpi6["scores_research"]
    + 0.30 * the_kpi6["scores_citations"]
    + 0.20 * the_kpi6["scores_international_outlook"]
)

kpi6_final = the_kpi6[
    [
        "university_id",
        "research_productivity_index"
    ]
]

final_df = final_df.merge(
    kpi6_final,
    on="university_id",
    how="left"
)

# ============================================================
# SOURCE + YEAR INFORMATION
# ============================================================

final_df["overview_year"] = 2025
final_df["research_year"] = 2024
final_df["student_year"] = 2023

final_df["global_ranking_score_source"] = "QS 2025 Overall Score"
final_df["research_impact_score_source"] = (
    "QS 2025 Citations per Faculty Score"
)
final_df["faculty_to_student_ratio_source"] = (
    "WUR 2023 Students per Staff"
)
final_df["international_student_percentage_source"] = (
    "WUR 2023 International Student Percentage"
)
final_df["academic_reputation_score_source"] = (
    "QS 2025 Academic Reputation Score"
)
final_df["research_productivity_index_source"] = (
    "THE 2024: 50% Research + 30% Citations + 20% International Outlook"
)

# ============================================================
# ROUND KPI VALUES
# ============================================================

kpi_columns = [
    "global_ranking_score",
    "research_impact_score",
    "faculty_to_student_ratio",
    "international_student_percentage",
    "academic_reputation_score",
    "research_productivity_index"
]

for column in kpi_columns:
    final_df[column] = pd.to_numeric(
        final_df[column],
        errors="coerce"
    ).round(2)

# ============================================================
# REMOVE INVALID INFINITE VALUES
# ============================================================

final_df = final_df.replace(
    [np.inf, -np.inf],
    np.nan
)

# ============================================================
# FINAL COLUMN ORDER
# ============================================================

final_df = final_df[
    [
        "university_id",
        "university_name",
        "country_id",
        "country_name",
        "region",

        "overview_year",
        "research_year",
        "student_year",

        "global_ranking_score",
        "global_ranking_score_source",

        "research_impact_score",
        "research_impact_score_source",

        "faculty_to_student_ratio",
        "faculty_to_student_ratio_source",

        "international_student_percentage",
        "international_student_percentage_source",

        "academic_reputation_score",
        "academic_reputation_score_source",

        "research_productivity_index",
        "research_productivity_index_source"
    ]
]

# ============================================================
# VALIDATION
# ============================================================

print("===== FINAL EDUCATION KPI DATASET =====")
print("Total universities:", len(final_df))
print("Total columns:", len(final_df.columns))
print("Duplicate University IDs:",
      final_df["university_id"].duplicated().sum())

print("\nKPI availability:")

for column in kpi_columns:
    print(
        column, ":",
        final_df[column].notna().sum(),
        "available /",
        final_df[column].isna().sum(),
        "missing"
    )

# ============================================================
# SAVE TO EXCEL
# ============================================================

output_file = os.path.join(
    PROCESSED_PATH,
    "university_final_dataset.xlsx"
)

final_df.to_excel(
    output_file,
    index=False
)

print("\nSaved successfully:")
print(output_file)

print("\nSample:")
print(final_df.head(5).to_string(index=False))
```
