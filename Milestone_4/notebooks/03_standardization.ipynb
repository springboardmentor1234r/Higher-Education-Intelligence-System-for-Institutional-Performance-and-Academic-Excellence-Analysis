# EduVision_DV — Standardization
Creates normalized university and country keys for controlled integration.

```python
import pandas as pd, re
def normalize_name(x):
    x=re.sub(r"^the\\s+","",str(x).lower().strip())
    return re.sub(r"[^a-z0-9]","",x)
du=pd.read_csv("../data_final/dim_university.csv")
du["normalized_name"]=du["university_name"].map(normalize_name)
print(du[["university_id","university_name","normalized_name"]].head())
```
