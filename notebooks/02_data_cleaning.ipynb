Cleans duplicates, standardizes missing values and parses numeric fields.

```python
import pandas as pd, re
from pathlib import Path
ROOT=Path("..")
for path in (ROOT/"data_final").glob("*_cleaned.csv"):
    df=pd.read_csv(path)
    print(path.name, "rows:",len(df), "duplicates:",df.duplicated().sum(), "missing:",df.isna().sum().sum())
```
